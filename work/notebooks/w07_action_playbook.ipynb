## 1. Ranked actions + reason codes

*The queue: what to do first, and why, in words a human trusts.*

```python
'''
Reason codes: CLICK_DECLINE, IMPRESSION_DECLINE, and POSITION_WORSEN.
CLICK_DECLINE, IMPRESSION_DECLINE: clicks / impression is predicted by the model to decrease at least 15% relative to last 30 days (days 31-60).
POSITION_WORSEN: average position is predicted by the model to increase at least 15% relative to last 30 days.
The pages are ranked based on how many of these 3 flags a page has. This shows how many of its metrics are at risk.
Update the pages with all 3 flags first, then 2 flags, and then 1 flag.
'''
```

## 2. Intended use and limits

*Who uses this, for what — and where it stops being valid.*

```python
'''
SEO professionals use this to see which pages to prioritize updating and what component to update.
For example, if clicks is predicted to decrease, they might update the title to be more attractive.
If position is predicted to go deeper, they might update the content to show the most recent information.
It can also be used by non SEO professionals, for example business planning.
For example, if the clicks for a page selling a product is predicted to increase, they might consider increasing the inventory for this product.
Limits: it does not say what is the reason for its predicted values for clicks, impressions, or position.
It tends to overestimate smaller values and underestimate larger values for each clicks, impressions, or position.
'''
```

## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*

```python
'''
The reviewer should look if there was a festive season in the near past or future that could cause the metrics to be very high.
Since the model is not given any exact dates, its prediction does not reflect the impact of festive seasons.
They should also check if the demand for such pages have substantially decreased.
In this case, while the model's prediction is likely accurate, trying to fix this page may not be the best solution.
Updating the content or allocating resources should never be automated before human review.
'''
```

## 4. Monitoring / retrain triggers

*What would tell you the recommendations went stale?*

```python
'''
Monitor the r2 score of the model over time as new actual outcomes become available.
If the score is consistently and significantly lower than the initial scores, the recommendations likely went stale.
This might be due to major changes in search engine algorithms.
If this occurs, feed in the new data to retrain the model.
Do not retrain if the r2 score is lower for a very short time, it might be due to a temporary peak season.
'''
```

```python
import joblib
model1 = joblib.load("work/outputs/model1.pkl")
model2 = joblib.load("work/outputs/model2.pkl")
model3 = joblib.load("work/outputs/model3.pkl")
X_test = joblib.load("work/outputs/X_test.pkl")
y_test1 = joblib.load("work/outputs/y_test1.pkl")
y_test2 = joblib.load("work/outputs/y_test2.pkl")
y_test3 = joblib.load("work/outputs/y_test3.pkl")
```

```python
import pandas as pd
table = pd.DataFrame({
"Hyperparameter": ["max_iter", "learning_rate", "min_samples_leaf", "l2_regularization"],
    "Clicks model": [model1.max_iter, model1.learning_rate, model1.min_samples_leaf, model1.l2_regularization],
    "Impressions model": [model2.max_iter, model2.learning_rate, model2.min_samples_leaf, model2.l2_regularization],
    "Position model": [model3.max_iter, model3.learning_rate, model3.min_samples_leaf, model3.l2_regularization],
})
table.head()
```

```python
import csv
pred1 = model1.predict(X_test)
pred2 = model2.predict(X_test)
pred3 = model3.predict(X_test)
headers = ['clicks_1_15', 'clicks_16_30', 'clicks_31_45', 'clicks_46_60', 
           'impressions_1_15', 'impressions_16_30', 'impressions_31_45', 'impressions_46_60',
           'position_1_15', 'position_16_30', 'position_31_45', 'position_46_60', 
           'word_count', 'update_date', 'create_date', 'main_intent',
           'pred_clicks', 'pred_impressions', 'pred_position',
           'true_clicks', 'true_impressions', 'true_position',
           'clicks_decline', 'impressions_decline', 'position_worsen',
           'clicks_decline_baseline', 'impressions_decline_baseline', 'position_worsen_baseline']
data = []
click_model = impression_model = position_model = click_baseline = impression_baseline = position_baseline = 0
click_model_correct = impression_model_correct = position_model_correct = click_baseline_correct = impression_baseline_correct = position_baseline_correct = 0
for i in range(len(X_test)):
    arr = X_test[i].tolist()
    arr.extend([pred1[i], pred2[i], pred3[i]])
    arr.extend([y_test1[i], y_test2[i], y_test3[i]])
    arr.append(int(pred1[i] <= 0.85 * (X_test[i][2] + X_test[i][3])))
    arr.append(int(pred2[i] <= 0.85 * (X_test[i][6] + X_test[i][7])))
    pos1 = (X_test[i][8] * X_test[i][4] + X_test[i][9] * X_test[i][5]) / (X_test[i][4] + X_test[i][5])
    pos2 = (X_test[i][10] * X_test[i][6] + X_test[i][11] * X_test[i][7]) / (X_test[i][6] + X_test[i][7])
    arr.append(int(pred3[i] >= 1.15 * pos2))
    arr.append(int(X_test[i][2] + X_test[i][3] <= 0.85 * (X_test[i][0] + X_test[i][1])))
    arr.append(int(X_test[i][6] + X_test[i][7] <= 0.85 * (X_test[i][4] + X_test[i][5])))
    arr.append(int(pos2 >= 1.15 * pos1))
    if arr[22]:
        click_model += 1
        if arr[19] <= 0.9 * (X_test[i][2] + X_test[i][3]): click_model_correct += 1
    if arr[23]:
        impression_model += 1
        if arr[20] <= 0.9 * (X_test[i][4] + X_test[i][5]): impression_model_correct += 1
    if arr[24]:
        position_model += 1
        if arr[21] >= 1.1 * pos2: position_model_correct += 1
    if arr[25]:
        click_baseline += 1
        if arr[19] <= 0.9 * (X_test[i][2] + X_test[i][3]): click_baseline_correct += 1
    if arr[26]:
        impression_baseline += 1
        if arr[20] <= 0.9 * (X_test[i][4] + X_test[i][5]): impression_baseline_correct += 1
    if arr[27]:
        position_baseline += 1
        if arr[21] >= 1.1 * pos2: position_baseline_correct += 1
    data.append(arr)
with open('work/outputs/result.csv', mode='w', newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(headers)
    writer.writerows(data)
with open('work/outputs/flag_analysis', mode='w') as f:
    f.write(f'''Model precision:\nClicks: {click_model_correct / click_model} 
Impressions: {impression_model_correct / impression_model} 
Position: {position_model_correct / position_model}\n''')
    f.write(f'''Baseline precision:\nClicks: {click_baseline_correct / click_baseline},
Impressions: {impression_baseline_correct / impression_baseline}, 
Position: {position_baseline_correct / position_baseline}\n''')
    f.write(f'''Model flags:\nClicks {click_model}, Impressions: {impression_model}, Position: {position_model}\n''')
    f.write(f'''Baseline flags:\nClicks {click_baseline}, Impressions: {impression_baseline}, Position: {position_baseline}\n''')
```
